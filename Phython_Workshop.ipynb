**Print number table**

```python
num=input("which number table you want to print")
num=int(num)
i=1
while(i<=10):
  print(f"{num} * {i} = {num*i}")
  i=i+1
```

**Break Statement**

```python
num=input("which number table you want to print")
num=int(num)
i=1
while(i<=10):
  print(f"{num} * {i} = {num*i}")
  if(1==5):
    break
  i=i+1
```

```python
List_of_Books=["python", "java", "C", "C++"]
book=input("enter book you want in library")
book_no=0 #start from index 0
while(book_no < len(List_of_Books)):
  if(book==List_of_Books[book_no]):
    print("yes, book available")
    break
  book_no=book_no+1

```
